### Chronomètre

```python
class Chronometre:
    def __init__(self, fenetre):
        self.fenetre = fenetre
        self.fenetre.title("Chronomètre")
        self.fenetre.geometry("350x200")

        self.temps = 0
        self.en_cours = False
        self.depart = None
        self.splits = []

        self.affichage = tk.Label(fenetre,
                                  text="00:00:00",
                                  font=("Arial", 40))
        self.affichage.pack(pady=20)

        cadre = tk.Frame(fenetre)
        cadre.pack()

        tk.Button(cadre, text="Démarrer",
                  command=self.demarrer).pack(side="left", padx=5)

        tk.Button(cadre, text="Pause",
                  command=self.pause).pack(side="left", padx=5)

        tk.Button(cadre, text="Réinitialiser",
                  command=self.reset).pack(side="left", padx=5)

        fenetre.bind("<e>", self.split)

    def demarrer(self):
        if not self.en_cours:
            self.depart = time.time() - self.temps
            self.en_cours = True
            self.actualiser()

    def pause(self):
        self.en_cours = False
        print(self.splits)

    def reset(self):
        self.en_cours = False
        self.temps = 0
        self.affichage.config(text="00:00:00")

    def actualiser(self):
        if self.en_cours:
            self.temps = time.time() - self.depart

            minutes = int((self.temps % 3600) // 60)
            secondes = int(self.temps % 60)
            centiemes = int((self.temps*100)%100)

            global timer 
            timer = f"{minutes:02d}:{secondes:02d}:{centiemes:02d}"
            self.affichage.config(text=timer)

            self.fenetre.after(50, self.actualiser)

    def split(self, event):
        self.splits.append(timer)
        print(timer)


fenetre = tk.Tk()
app = Chronometre(fenetre)
fenetre.mainloop()
```
